```python
silver_table = "workspace.silver.yellow_taxi_trips"

spark.sql("""
CREATE OR REPLACE TABLE workspace.gold.monthly_total_amount_avg AS
SELECT
    pickup_year,
    pickup_month,
    ROUND(
        AVG(total_amount),
        2
    ) AS avg_total_amount
FROM workspace.silver.yellow_taxi_trips
GROUP BY
    pickup_year,
    pickup_month
ORDER BY
    pickup_month
""")

spark.sql("""
CREATE OR REPLACE TABLE workspace.gold.executive_summary AS
SELECT
    COUNT(*) AS total_trips,
    ROUND(AVG(total_amount), 2) AS avg_total_amount,
    ROUND(AVG(passenger_count), 2) AS avg_passenger_count,
    MIN(pickup_datetime) AS first_trip,
    MAX(pickup_datetime) AS last_trip
FROM workspace.silver.yellow_taxi_trips
""")

display(
    spark.table("workspace.gold.executive_summary")
)
```
