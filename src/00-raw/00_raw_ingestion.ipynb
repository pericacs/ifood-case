```python
# Databricks notebook source

# COMMAND ----------

import requests

# COMMAND ----------

taxi_type = "yellow"
year = 2023
months = [1, 2, 3, 4, 5]

source_base_url = "https://d37ci6vzurychx.cloudfront.net/trip-data"

landing_zone_path = "/Volumes/workspace/raw/landing_zone/yellow_taxi"

# COMMAND ----------

for month in months:
    month_str = str(month).zfill(2)

    file_name = f"{taxi_type}_tripdata_{year}-{month_str}.parquet"
    source_url = f"{source_base_url}/{file_name}"

    target_dir = f"{landing_zone_path}/year={year}/month={month_str}"
    target_file = f"{target_dir}/{file_name}"

    dbutils.fs.mkdirs(target_dir)

    print(f"Baixando arquivo: {source_url}")
    print(f"Salvando em: {target_file}")

    response = requests.get(source_url, timeout=300)

    if response.status_code != 200:
        raise Exception(f"Erro ao baixar arquivo {file_name}. Status: {response.status_code}")

    with open(target_file, "wb") as file:
        file.write(response.content)

    print(f"Arquivo salvo com sucesso: {file_name}")

# COMMAND ----------

display(dbutils.fs.ls(f"{landing_zone_path}/year=2023"))
```
